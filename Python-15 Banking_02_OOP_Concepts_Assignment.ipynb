**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
class Transfer:
    def __init__(self, from_account, to_account, amount):
        self.from_account = from_account
        self.to_account = to_account
        self.amount = amount
        self.status = "Pending"

    def execute(self):
        if self.amount <= 0:
            self.status = "Failed: Invalid amount"
            return self.status

        if self.from_account.balance >= self.amount:
            self.from_account.withdraw(self.amount)
            self.to_account.deposit(self.amount)
            self.status = "Success"
        else:
            self.status = "Failed: Insufficient funds"
        return self.status

    def __str__(self):
        return (f"Transfer {self.amount} from {self.from_account.account_number} "
                f"to {self.to_account.account_number} - Status: {self.status}")



if __name__ == "__main__":

    acc_from = Account("ACC-1003", "Suvarna", 5000)
    acc_to = Account("ACC-2001", "Ravi", 2000)


    transfer1 = Transfer(acc_from, acc_to, 1500)


    print(transfer1.execute())



    print(transfer1)



    print(acc_from.get_balance())

    print(acc_to.get_balance())


```

### Q8. Protect a amount with a property for banking case 1; use sample reference `ACC-1008` and explain the result.

```python
# Write your solution here
class Account:
    branch_name = "MG Road Branch"

    def __init__(self, account_number, holder_name, balance=0.0):
        self.account_number = account_number
        self.holder_name = holder_name
        self._balance = balance


    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = amount

    def deposit(self, amount):
        if amount > 0:
            self.balance = self._balance + amount
            return f"Deposited {amount}. New balance: {self.balance}"
        else:
            return "Deposit amount must be positive."

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance = self._balance - amount
            return f"Withdrew {amount}. New balance: {self.balance}"
        else:
            return "Insufficient funds or invalid amount."

    def __str__(self):
        return (f"Account[{self.account_number}] | Holder: {self.holder_name} | "
                f"Balance: {self.balance} | Branch: {Account.branch_name}")



if __name__ == "__main__":
    acc8 = Account("ACC-1008", "Suvarna", 9000)


    print(acc8.deposit(1000))



    print(acc8.withdraw(4000))



    print(acc8.balance)

    try:
        acc8.balance = -500
    except ValueError as e:
        print(e)

```

### Q14. Model a loyaltyaccount for banking case 1; use sample reference `ACC-1014` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
class LoyaltyAccount:
    def __init__(self, account_number, holder_name, balance=0.0):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance
        self.points = 0

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            self.points += int(amount // 100)
            return f"Deposited {amount}. New balance: {self.balance}. Points earned: {self.points}"
        else:
            return "Deposit amount must be positive."

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount

            self.points += int(amount // 200)
            return f"Withdrew {amount}. New balance: {self.balance}. Points earned: {self.points}"
        else:
            return "Insufficient funds or invalid amount."

    def redeem_points(self, points_to_redeem):
        if 0 < points_to_redeem <= self.points:
            self.points -= points_to_redeem
            return f"Redeemed {points_to_redeem} points. Remaining points: {self.points}"
        else:
            return "Not enough points to redeem."

    def __str__(self):
        return (f"LoyaltyAccount[{self.account_number}] | Holder: {self.holder_name} | "
                f"Balance: {self.balance} | Points: {self.points}")



if __name__ == "__main__":
    acc14 = LoyaltyAccount("ACC-1014", "Suvarna", 5000)


    print(acc14.deposit(1200))



    print(acc14.withdraw(800))

    print(acc14.redeem_points(10))

    print(acc14)

```

### Q93. Build a state machine for transfer status for banking case 3; use sample reference `ACC-1093` and explain the result.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here

class TransferStateMachine:
    def __init__(self, reference_id):
        self.reference_id = reference_id
        self.state = "INITIATED"

    def process(self):
        if self.state == "INITIATED":
            self.state = "VALIDATING"
            return f"{self.reference_id}: Transfer moved to VALIDATING"
        elif self.state == "VALIDATING":
            self.state = "PROCESSING"
            return f"{self.reference_id}: Transfer moved to PROCESSING"
        elif self.state == "PROCESSING":
            self.state = "SETTLED"
            return f"{self.reference_id}: Transfer moved to SETTLED"
        else:
            return f"{self.reference_id}: Transfer already {self.state}"

    def fail(self, reason="Unknown"):
        if self.state in ["VALIDATING", "PROCESSING"]:
            self.state = "FAILED"
            return f"{self.reference_id}: Transfer FAILED due to {reason}"
        else:
            return f"{self.reference_id}: Cannot fail from {self.state}"

    def rollback(self):
        if self.state == "FAILED":
            self.state = "ROLLED_BACK"
            return f"{self.reference_id}: Transfer rolled back"
        else:
            return f"{self.reference_id}: Rollback not allowed from {self.state}"

    def __repr__(self):
        return f"Transfer[{self.reference_id}] | Current State: {self.state}"

transfer = TransferStateMachine("ACC-1093")

print(transfer)                # INITIATED
print(transfer.process())      # VALIDATING
print(transfer.process())      # PROCESSING
print(transfer.fail("Perishable account mismatch"))  # FAILED
print(transfer.rollback())     # ROLLED_BACK
print(transfer)                # Final state
```
