### Q33. Use polymorphic payment processors (retail variant 1).

```python
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self,amount): ...
class Card(Payment):
    def pay(self,amount): return f'Card charged {amount:.2f}'
print(Card().pay(250))
```

**Explanation:** The model keeps data and relevant behavior together. Notice where inheritance expresses an “is-a” relationship and composition expresses a “has-a” relationship; public methods protect internal state.

### Q34. Compose order from orderline objects (retail variant 1).

```python
class OrderLine:
    def __init__(self,sku,qty,price): self.sku,self.qty,self.price=sku,qty,price
    def total(self): return self.qty*self.price
class Order:
    def __init__(self,lines): self.lines=list(lines)
    def total(self): return sum(x.total() for x in self.lines)
print(Order([OrderLine('A',2,50)]).total())
```

**Explanation:** The model keeps data and relevant behavior together. Notice where inheritance expresses an “is-a” relationship and composition expresses a “has-a” relationship; public methods protect internal state.

### Q93. Build a state machine for order status (retail variant 3).

```python
class Order:
    allowed={'new':{'paid'},'paid':{'packed','refunded'},'packed':{'shipped'},'shipped':set()}
    def __init__(self): self.state='new'
    def move(self,new):
        if new not in self.allowed[self.state]: raise ValueError(f'{self.state} -> {new} not allowed')
        self.state=new
o=Order(); o.move('paid'); o.move('packed'); print(o.state)
```

**Explanation:** The model keeps data and relevant behavior together. Notice where inheritance expresses an “is-a” relationship and composition expresses a “has-a” relationship; public methods protect internal state.
