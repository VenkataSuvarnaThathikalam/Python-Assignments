Assignment 32: Function Returning a Dictionary

Write a function named analyze_text().

Requirements
Accept a string.
Return a dictionary containing:
Total characters
Total words
Total vowels
Total digits
Total spaces
Display every result clearly.

```python
def analyze_text(text):
    char_count = len(text)
    word_count = len(text.split())
    vowel_count = 0
    digit_count = 0
    space_count = 0

    vowels = "aeiouAEIOU"
    for char in text:
        if char in vowels:
            vowel_count += 1
        elif char.isdigit():
            digit_count += 1
        elif char.isspace():
            space_count += 1
    result = {
        "Total characters": char_count,
        "Total words": word_count,
        "Total vowels": vowel_count,
        "Total digits": digit_count,
        "Total spaces": space_count
    }

    return result
text = "Hello World"
result = analyze_text(text)
for key, value in result.items():
    print(f"{key}: {value}")
```

Assignment 35: Fibonacci Function

Write a function named generate_fibonacci().

Requirements
Accept the number of terms.
Return the complete Fibonacci series as a list.
Handle zero and negative term counts.
Do not use recursion in this assignment.

```python
def generate_fibonacci(n):
  if n<=0:
    return "Invalid number of terms"
  fibonacci_series=[0,1]
  while len(fibonacci_series)<n:
    next_number=fibonacci_series[-1]+fibonacci_series[-2]
    fibonacci_series.append(next_number)
  return fibonacci_series
n=int(input("Enter the number of terms:"))
result=generate_fibonacci(n)
print("Fibonacci series:",result )
```

```python
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }

]
def display_employees():
    for employee in employees:
        print(employee)
display_employees()
def display_employees_names():
    employee_names = list(map(lambda employee: employee["name"], employees))
    print(employee_names)
display_employees_names()
def filter_high_performers():
    high_performers = list(filter(lambda employee: employee["performance_score"] >= 80, employees))
    print(high_performers)
filter_high_performers()
def filter_by_department(department):
    department_employees = list(filter(lambda employee: employee["department"] == department, employees))
    print("Department Employees: ", department_employees)
filter_by_department("IT")
def calculate_revised_salaries():
    def calculate_salary_for_employee(employee):
        if employee["performance_score"] >= 90:
            return employee["salary"] * 1.15
        elif employee["performance_score"] >= 80:
            return employee["salary"] * 1.10
        elif employee["performance_score"] >= 70:
            return employee["salary"] * 1.05
        else:
            return employee["salary"]
    revised_salaries = list(map(calculate_salary_for_employee, employees))
    print("Revised Salaries: ", revised_salaries)
    return revised_salaries # Return the list
calculate_revised_salaries()
def calculate_total_salary_expense():
    current_total_salary = sum(employee["salary"] for employee in employees)
    revised_salaries_list = calculate_revised_salaries()
    revised_total_salary = sum(revised_salaries_list)
    total_additional_salary_expense = revised_total_salary - current_total_salary
    print("Current Total Salary: ", current_total_salary)
    print("Revised Total Salary: ", revised_total_salary)
    print("Total Additional Salary Expense: ", total_additional_salary_expense)
calculate_total_salary_expense()
def find_highest_paid_employee():
    highest_paid_employee = max(employees, key=lambda employee: employee["salary"])
    print("Highest Paid Employee: ", highest_paid_employee)
find_highest_paid_employee()
def sort_employees(key):
    sorted_employees = sorted(employees, key=lambda employee: employee[key])
    print("Sorted Employees: ", sorted_employees)
sort_employees("name")
sort_employees("salary")
sort_employees("performance_score")
sort_employees("experience")
def create_bonus_calculator(bonus_percentage):
    def calculate_bonus(salary):
        return salary * (bonus_percentage / 100)
    return calculate_bonus
calculate_bonus_5 = create_bonus_calculator(5)
calculate_bonus_10 = create_bonus_calculator(10)
calculate_bonus_15 = create_bonus_calculator(15)
def generate_department_report(department):
    department_employees = list(filter(lambda employee: employee["department"] == department, employees))
    num_employees = len(department_employees)
    total_salary = sum(employee["salary"] for employee in department_employees)
    average_salary = total_salary / num_employees
    average_performance_score = sum(employee["performance_score"] for employee in department_employees) / num_employees
    report = {
                "Department": department,
                "Number of Employees": num_employees,
                "Total Salary": total_salary,
                "Average Salary": average_salary,
                "Average Performance Score": average_performance_score
    }
    return report
generate_department_report("IT")
def menu():
    print("1. Display all employees")
    print("2. Display employee names")
    print("3. Display high performers")
    print("4. Search by department")
    print("5. Calculate revised salaries")
    print("6. Calculate total salary expense")
    print("7. Find highest-paid employee")
    print("8. Sort employees")
    print("9. Generate department report")
    print("10. Exit")
while True:
    menu()
    choice = input("Enter your choice: ")
    if choice == "1":
        display_employees()
    elif choice == "2":
        display_employees_names()
    elif choice == "3":
        filter_high_performers()
    elif choice == "4":
        department = input("Enter department name: ")
        filter_by_department(department)
    elif choice == "5":
        calculate_revised_salaries()
    elif choice == "6":
        calculate_total_salary_expense()
    elif choice == "7":
        find_highest_paid_employee()
    elif choice == "8":
        sort_key = input("Enter sorting key (name, salary, performance_score, experience): ")
        sort_employees(sort_key)
    elif choice == "9":
        department = input("Enter department name: ")
        report = generate_department_report(department)
        print(report)
    elif choice == "10":
        print("Exiting the program.")
        break
    else:
        print("Invalid choice. Please try again.")
```
