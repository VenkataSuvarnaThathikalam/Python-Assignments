```python
# Write your solution here
# Banking Case 1: Process transfers while isolating bad rows

class InvalidTransactionError(Exception):
    """Raised when a transaction row is invalid."""
    def __init__(self, account_ref, reason):
        super().__init__(f"Account {account_ref} - {reason}")
        self.account_ref = account_ref
        self.reason = reason


def process_transfers(transfers):

    good_rows = []
    bad_rows = []

    for row in transfers:
        try:
            account_ref = row.get("account_ref")
            balance = row.get("balance")
            amount = row.get("amount")

            # Validate row
            if account_ref is None or balance is None or amount is None:
                raise InvalidTransactionError(account_ref or "UNKNOWN", "Missing required fields")
            if amount <= 0:
                raise InvalidTransactionError(account_ref, "Amount must be positive")
            if amount > balance:
                raise InvalidTransactionError(account_ref, "Insufficient funds")

            # Process valid transfer
            balance -= amount
            good_rows.append({
                "account_ref": account_ref,
                "operation": "TRANSFER",
                "status": "SUCCESS",
                "remaining_balance": balance
            })

        except InvalidTransactionError as e:
            bad_rows.append({
                "account_ref": row.get("account_ref"),
                "operation": "TRANSFER",
                "status": f"FAILED - {e.reason}"
            })

    return {"good_rows": good_rows, "bad_rows": bad_rows}


# Example usage with ACC-1033
transfers = [
    {"account_ref": "ACC-1033", "balance": 500.0, "amount": 200.0},  # valid
    {"account_ref": "ACC-1033", "balance": 300.0, "amount": 400.0},  # insufficient funds
    {"account_ref": "ACC-1033", "balance": 250.0, "amount": -50.0},  # invalid amount
]

result = process_transfers(transfers)
print(result)
```

```python


import random
import time

class PaymentGatewayError(Exception):

    def __init__(self, account_ref, reason):
        super().__init__(f"[{account_ref}] {reason}")
        self.account_ref = account_ref
        self.reason = reason


def simulate_gateway_call():

    if random.choice([True, False]):  # 50% chance of failure
        raise ConnectionError("Transient gateway failure.")
    return "Payment processed successfully."


def bounded_exponential_retry(account_ref: str, amount: float, max_retries: int = 5, base_delay: float = 1.0, max_delay: float = 16.0):

    attempt = 0
    while attempt < max_retries:
        try:
            result = simulate_gateway_call()
            return {
                "account_ref": account_ref,
                "operation": "PAYMENT",
                "status": "SUCCESS",
                "amount": amount,
                "message": result,
                "attempts": attempt + 1
            }
        except ConnectionError as e:
            attempt += 1
            delay = min(base_delay * (2 ** (attempt - 1)), max_delay)
            print(f"Attempt {attempt} failed: {e}. Retrying in {delay} seconds...")
            time.sleep(delay)

    raise PaymentGatewayError(account_ref, f"Payment failed after {max_retries} retries.")


try:
    result = bounded_exponential_retry("ACC-1074", 1000.0)
    print(result)
except PaymentGatewayError as e:
    print({
        "account_ref": e.account_ref,
        "status": "FAILED",
        "error_type": "PaymentGatewayError",
        "reason": e.reason
    })
```

```python




class TransactionError(Exception):
    """Custom error for transaction failures in case 3."""
    def __init__(self, account_ref, reason):
        super().__init__(f"[{account_ref}] Transaction failed: {reason}")
        self.account_ref = account_ref
        self.reason = reason


class TransactionContext:

    def __init__(self, account_ref, balance):
        self.account_ref = account_ref
        self.original_balance = balance
        self.balance = balance

    def __enter__(self):
        print(f"Starting transaction for {self.account_ref}")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type:
            # Rollback on error
            self.balance = self.original_balance
            print(f"Transaction rolled back for {self.account_ref}: {exc_value}")
            return True  # suppress exception
        else:
            # Commit success
            print(f"Transaction committed for {self.account_ref}")
            return False


def process_transaction_case3(account_ref: str, balance: float, amount: float, transaction_type: str):

    with TransactionContext(account_ref, balance) as txn:
        if transaction_type not in ["WITHDRAWAL", "DEPOSIT"]:
            raise TransactionError(account_ref, "Unsupported transaction type")
        if amount <= 0:
            raise TransactionError(account_ref, "Amount must be positive")
        if transaction_type == "WITHDRAWAL" and amount > txn.balance:
            raise TransactionError(account_ref, "Insufficient funds")

        # Apply transaction
        if transaction_type == "WITHDRAWAL":
            txn.balance -= amount
        elif transaction_type == "DEPOSIT":
            txn.balance += amount

        return {
            "account_ref": account_ref,
            "operation": transaction_type,
            "status": "SUCCESS",
            "requested_amount": amount,
            "remaining_balance": txn.balance
        }


# Example usage with ACC-1095
result = process_transaction_case3("ACC-1095", balance=700.0, amount=900.0, transaction_type="WITHDRAWAL")
print(result)
```

### Q97. Make batch settlement failure-isolated for banking case 3; use sample reference `ACC-1097` and explain the result.

```python

class SettlementError(Exception):

    def __init__(self, account_ref, reason):
        super().__init__(f"[{account_ref}] Settlement failed: {reason}")
        self.account_ref = account_ref
        self.reason = reason


def process_settlement_row_case3(account_ref: str, balance: float, amount: float, transaction_type: str):

    if transaction_type not in ["WITHDRAWAL", "DEPOSIT", "TRANSFER"]:
        raise SettlementError(account_ref, "Unsupported transaction type")
    if amount <= 0:
        raise SettlementError(account_ref, "Amount must be positive")
    if transaction_type == "WITHDRAWAL" and amount > balance:
        raise SettlementError(account_ref, "Insufficient funds")

    # Success path
    if transaction_type == "WITHDRAWAL":
        balance -= amount
    elif transaction_type == "DEPOSIT":
        balance += amount
    elif transaction_type == "TRANSFER":
        if amount > balance:
            raise SettlementError(account_ref, "Insufficient funds for transfer")
        balance -= amount

    return {
        "account_ref": account_ref,
        "operation": transaction_type,
        "status": "SUCCESS",
        "requested_amount": amount,
        "remaining_balance": balance
    }


def batch_settlement_case3(transactions):

    good_rows = []
    bad_rows = []

    for tx in transactions:
        try:
            result = process_settlement_row_case3(
                tx["account_ref"], tx["balance"], tx["amount"], tx["transaction_type"]
            )
            good_rows.append(result)
        except SettlementError as e:
            bad_rows.append({
                "account_ref": e.account_ref,
                "operation": tx["transaction_type"],
                "status": "FAILED",
                "error_type": "SettlementError",
                "reason": e.reason,
                "requested_amount": tx["amount"],
                "available_balance": tx["balance"]
            })

    return {"good_rows": good_rows, "bad_rows": bad_rows}


# Example usage with ACC-1097
transactions = [
    {"account_ref": "ACC-1097", "balance": 2000.0, "amount": 500.0, "transaction_type": "WITHDRAWAL"},  # valid
    {"account_ref": "ACC-1097", "balance": 300.0, "amount": 600.0, "transaction_type": "WITHDRAWAL"},   # insufficient funds
    {"account_ref": "ACC-1097", "balance": 1000.0, "amount": -200.0, "transaction_type": "DEPOSIT"},    # invalid amount
]

result = batch_settlement_case3(transactions)
print(result)
```

### Q99. Design idempotent recovery after timeout for banking case 3; use sample reference `ACC-1099` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python


import uuid
import random

class TimeoutError(Exception):

    def __init__(self, account_ref, reason):
        super().__init__(f"[{account_ref}] Timeout: {reason}")
        self.account_ref = account_ref
        self.reason = reason



processed_transactions = {}


def simulate_gateway_call(transaction_id):

    if random.choice([True, False]):  # 50% chance of timeout
        raise TimeoutError(transaction_id, "Settlement gateway timeout in case 3")
    return "Transfer processed successfully."


def idempotent_transaction_case3(account_ref: str, amount: float, transaction_type: str):

    # Generate a unique transaction ID
    transaction_id = str(uuid.uuid4())

    try:
        # Check if transaction already processed
        if transaction_id in processed_transactions:
            return {
                "account_ref": account_ref,
                "operation": transaction_type,
                "status": "SUCCESS",
                "amount": amount,
                "message": "Transaction already processed (idempotent recovery)",
                "transaction_id": transaction_id
            }

        # Attempt gateway call
        message = simulate_gateway_call(transaction_id)

        # Mark transaction as processed
        processed_transactions[transaction_id] = {
            "account_ref": account_ref,
            "amount": amount,
            "operation": transaction_type,
            "status": "SUCCESS"
        }

        return {
            "account_ref": account_ref,
            "operation": transaction_type,
            "status": "SUCCESS",
            "amount": amount,
            "message": message,
            "transaction_id": transaction_id
        }

    except TimeoutError as e:
        return {
            "account_ref": account_ref,
            "operation": transaction_type,
            "status": "FAILED",
            "error_type": "TimeoutError",
            "reason": e.reason,
            "transaction_id": transaction_id,
            "idempotent_safe": True
        }



result = idempotent_transaction_case3("ACC-1099", 2500.0, "TRANSFER")
print(result)
```

### Q100. Build an exception-safe reconciliation pipeline for banking case 3; use sample reference `ACC-1100` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python


class ReconciliationError(Exception):

    def __init__(self, account_ref, step, reason):
        super().__init__(f"[{account_ref}] Reconciliation failed at {step}: {reason}")
        self.account_ref = account_ref
        self.step = step
        self.reason = reason


def validate_source(account_ref, source_data):
    if not source_data:
        raise ReconciliationError(account_ref, "validate_source", "Missing source data")
    return {"step": "validate_source", "status": "SUCCESS"}


def validate_target(account_ref, target_data):
    if not target_data:
        raise ReconciliationError(account_ref, "validate_target", "Missing target data")
    return {"step": "validate_target", "status": "SUCCESS"}


def compare_balances(account_ref, source_data, target_data):
    if source_data.get("balance") != target_data.get("balance"):
        raise ReconciliationError(account_ref, "compare_balances", "Balance mismatch")
    return {"step": "compare_balances", "status": "SUCCESS"}


def reconciliation_pipeline_case3(account_ref, source_data, target_data):

    steps = [
        lambda: validate_source(account_ref, source_data),
        lambda: validate_target(account_ref, target_data),
        lambda: compare_balances(account_ref, source_data, target_data)
    ]

    results = {"success": [], "failures": []}

    for step in steps:
        try:
            outcome = step()
            results["success"].append(outcome)
        except ReconciliationError as e:
            results["failures"].append({
                "account_ref": e.account_ref,
                "step": e.step,
                "status": "FAILED",
                "reason": e.reason
            })

    return results



source_data_example = {"balance": 8000.0}
target_data_example = {"balance": 7500.0}  # mismatch

result = reconciliation_pipeline_case3("ACC-1100", source_data_example, target_data_example)
print(result)
```
